# Clinical Note Summarizer — AI Demo
Uses LangChain + OpenAI GPT-4 to summarize clinical notes.
**Atlas monitors this notebook for AI library usage and data access patterns.**

```python
from langchain_openai import ChatOpenAI
from langchain.prompts import ChatPromptTemplate
from langchain.schema.output_parser import StrOutputParser
from transformers import pipeline
from huggingface_hub import hf_hub_download
import openai
import pandas as pd
```

```python
# Load a HuggingFace summarization model
summarizer = pipeline('summarization', model='facebook/bart-large-cnn')

# Download model artifacts from HuggingFace Hub
model_path = hf_hub_download(repo_id='distilbert/distilbert-base-uncased', filename='config.json')
```

```python
# LangChain pipeline with OpenAI GPT-4
llm = ChatOpenAI(model='gpt-4o', temperature=0)
prompt = ChatPromptTemplate.from_template('Summarize the following clinical note: {note}')
chain = prompt | llm | StrOutputParser()

# Example usage
result = chain.invoke({'note': 'Patient presented with...'})
print(result)
```
